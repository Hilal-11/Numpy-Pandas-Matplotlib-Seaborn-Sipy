```python
courses = pd.read_csv('./courses.csv')
matches = pd.read_csv('./matches.csv')
courses.head(10)
matches.head(10)
courses.sample(5)
courses.tail(5)
courses.info()
matches.info()

courses.describe()
courses.value_counts()
courses.sort_values(by='course_name', ascending=False)
courses.sort_values(by='course_name')
courses.sort_index()

courses.count()
courses.sum()

mean = courses['price'].mean()
median = courses['price'].median()
mode = courses['price'].mode()
variance = courses['price'].var()
std_dev = courses['price'].std()


print('\nMean:', mean)
print('\nMedian:', median)
print('\nMode:', mode)
print('\nVariance:', variance)
print('\nStandard Deviation:', std_dev)


min = courses['price'].min()
max = courses['price'].max()
print('\nMinimum:', min)
print('\nMaximum:', max)


```
