```python
data = np.random.randint(1, 30, size=(4, 4))
d2 = np.random.randint(1, 10, size=(1, 4))
result = data * d2
print("Array 1:\n", data)
print("\nArray 2:\n", d2)
print("\nBroadcasting result:\n", result)
```
