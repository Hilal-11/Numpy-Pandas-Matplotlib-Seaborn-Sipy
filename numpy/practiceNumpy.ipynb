```python

import time
import random
import numpy as np
 
 
def make_python_matrix(rows, cols):
    return [[random.random() for _ in range(cols)] for _ in range(rows)]
 
 
def python_matmul(A, B):
    rows_A, cols_A = len(A), len(A[0])
    rows_B, cols_B = len(B), len(B[0])
    result = [[0.0] * cols_B for _ in range(rows_A)]
    for i in range(rows_A):
        for j in range(cols_B):
            s = 0.0
            for k in range(cols_A):
                s += A[i][k] * B[k][j]
            result[i][j] = s
    return result
 
 
def python_elementwise_add(A, B):
    return [[A[i][j] + B[i][j] for j in range(len(A[0]))] for i in range(len(A))]
 
 
def python_elementwise_mul(A, B):
    return [[A[i][j] * B[i][j] for j in range(len(A[0]))] for i in range(len(A))]
 
 
def python_sum(A):
    total = 0.0
    for row in A:
        for val in row:
            total += val
    return total
 
 
def python_dot(a, b):
    return sum(x * y for x, y in zip(a, b))
 
 
def timed(func, *args, **kwargs):
    start = time.perf_counter()
    result = func(*args, **kwargs)
    end = time.perf_counter()
    return result, end - start
 
 
def compare_operation(name, python_time, numpy_time):
    speedup = python_time / numpy_time if numpy_time > 0 else float("inf")
    print(f"{name:<28} | Python: {python_time:>10.6f}s | NumPy: {numpy_time:>10.6f}s | Speedup: {speedup:>8.1f}x")



def run_comparison(n):
    print(f"\n{'='*90}")
    print(f"MATRIX SIZE: {n} x {n}")
    print(f"{'='*90}")
 
    # --- Setup ---
    A_py = make_python_matrix(n, n)
    B_py = make_python_matrix(n, n)
    A_np = np.array(A_py)
    B_np = np.array(B_py)
 
    vec_a_py = [random.random() for _ in range(n * n)]
    vec_b_py = [random.random() for _ in range(n * n)]
    vec_a_np = np.array(vec_a_py)
    vec_b_np = np.array(vec_b_py)
 
    # --- Matrix multiplication ---
    # NOTE: pure-python matmul is O(n^3) and gets very slow — skip for large n
    if n <= 200:
        _, t_py = timed(python_matmul, A_py, B_py)
    else:
        t_py = float("nan")
    _, t_np = timed(lambda: A_np @ B_np)
    compare_operation("Matrix Multiplication", t_py, t_np)
 
    # --- Element-wise addition ---
    _, t_py = timed(python_elementwise_add, A_py, B_py)
    _, t_np = timed(lambda: A_np + B_np)
    compare_operation("Element-wise Addition", t_py, t_np)
 
    # --- Element-wise multiplication ---
    _, t_py = timed(python_elementwise_mul, A_py, B_py)
    _, t_np = timed(lambda: A_np * B_np)
    compare_operation("Element-wise Multiplication", t_py, t_np)
 
    # --- Sum of all elements ---
    _, t_py = timed(python_sum, A_py)
    _, t_np = timed(lambda: A_np.sum())
    compare_operation("Sum of Elements", t_py, t_np)
 
    # --- Dot product (1D, size n*n) ---
    _, t_py = timed(python_dot, vec_a_py, vec_b_py)
    _, t_np = timed(lambda: np.dot(vec_a_np, vec_b_np))
    compare_operation("Dot Product (1D)", t_py, t_np)
    
    
if __name__ == "__main__":
    for size in [10, 100, 200, 500, 1000]:
        run_comparison(size)
        
```
